```sql
select * from student;

-- RANK() in window function


select * , rank() over(order by marks desc) as ranking_by_marks from student;

-- give the rank based on the student highest marsk with their respective branch

select * , rank() over(partition by branch order by marks desc) as branch_wise_ranking from student;

-- write the query to find the 1st rank student of every branch
select * from (select * , rank() over(partition by branch order by marks desc) as branch_wise_ranking from student) t 
where t.branch_wise_ranking = 1;

-- Rank students within each branch

select *, rank() over(partition by branch order by marks desc) as student_rank from student;

-- Find students who got 2nd rank in each branch
select * from (select * , rank() over(partition by branch order by marks desc) as student_rank from student) t 
where t.student_rank = 2;

--  Return only second position students.
select * from (select * , rank() over(order by marks desc) as student_rank from student) t 
where t.student_rank = 2

-- Rank all students overall
select  * ,rank() over(order by marks ) as overall_rank from student;

-- Find lowest scoring student in each branch

select * from (select *,rank() over(partition by branch order by marks) as low_score_student_rank from student) t
where t.low_score_student_rank = 1;

--  Rank in ascending order and pick lowest.

select * , rank() over(order by marks asc) as lowest_rank from student limit 1;

-- Find students whose rank is between 2 and 4
select * from (select *,rank() over(order by marks ) as student_rank from student) t  where t.student_rank between 2 and 4;

```
